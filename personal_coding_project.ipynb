For this coding project I will make a Rock-paper-scissors game where the participant will play a round against the computer that will have a random outcome.

First, I will import random to ... **explain**

```python
import random
```

Now I will make a variable, 'play1', which will contain my answer for the rock, paper, scissors game.

```python
play1 = str(input("Rock, Paper, or Scissors?")).capitalize()
```

By adding '.capitalize()' at the end of the string it makes it possible for the participant to write the front letter of rock, paper or scissors in either capital letters or lower case and it will automatically default to a capital letter in the system.

Now I will make a variable for the Bot that I'm playing against. This will be my opponents play in the game and will be shown as 'play2'. For this I will use the random module that will chose a random integer between 1 and 3, which will be either Rock, Paper or Scissors.

```python
play2 = random.randint(1,3)
```

Now I will define these numbers into words by using the 'elif' condition, which means that after it has checked the 'if' condition it is going to run the 'elif' and check that.  The 'elif' condition means that "if the outcome is this, do this, if the outcome is anything else, do this"

The quotation marks makes the outcome strings instead of integers, which... **(explain)**

```python
if play2 == 1:
    play2 = "Rock"

elif play2 == 2:
    play2 = "Paper"

elif play2 == 3:
    play2 = "Scissors"
```

Now I'm going to print what the bot has chosen to play in the given sitiuation, and I do this by using the 'print' function.

```python
print ("Bot: ", play2)
```

Next I will code statements that show that the bot and player either tie, won or lost the round, which I will do through another 'elif' condition.

```python
if play1 == play2:
    print("It's a tie!")

elif ((play1 == "Rock" and play2 == "Scissors") or (play1 == "Paper" and play2 == "Rock") or (play1 == "Scissors" and play2 == "Paper")):
    print("You win")

else: 
    print("You lose.")
```
